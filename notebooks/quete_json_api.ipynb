Etape 3 :

```python
import pandas as pd

data = pd.json_normalize(df, record_path = ['villes', 'restaurants'], meta = [['villes','nom'],['villes', 'region'],['villes', 'habitants']])
print(data)
```

Etape 4 :

```python
import statistics

data['note_moyenne'] = data['notes'].apply(lambda x: round(statistics.mean(x), 2))
print(data)
```

```python
print(f"La ville qui a la meilleure note moyenne est {data[data['note_moyenne'] == data['note_moyenne'].max()]['villes.nom'].iloc[0]}.")
```

## Challenge 2 :

Etape 1 :

```python
import requests

def interroger(url, params, headers=None):
    try:
        r = requests.get(url, params=params, headers=headers, timeout=10)
        r.raise_for_status()
        return r.json()
    except requests.exceptions.Timeout:
        print("Trop lent, on abandonne cette requête")
    except requests.exceptions.HTTPError as e:
        print("Le serveur a répondu une erreur :", e)
    except requests.exceptions.RequestException as e:
        print("Problème réseau :", e)
    return None
```

```python
headers = {"User-Agent": "bootcamp-simplon-da/1.0 (jordanpiade@gmail.com)"}
url = "https://api-adresse.data.gouv.fr/search/"
r = requests.get(url, params={"q": "20 avenue de Ségur 75007 Paris"}, headers=headers)

print(r.status_code)
print(r.json()["features"][0]["properties"]["label"])
print(r.json()["features"][0]["properties"]["score"])
print(r.json())
```

Etape 2 :

```python
adresses = [
    "20 avenue de Ségur 75007 Paris",
    "Grand Place, Lille",
    "1 rue de la Paix, Paris",
    "place Bellecour, Lyon",
    "zzzzz qwerty 99999",
]

liste = []
for adresse in adresses:
    reponse = interroger(url, {"q": adresse, "limit": 1})
    if len(reponse["features"]) > 0:
        liste.append({  "adresse" : adresse,
                        "label" : reponse["features"][0]["properties"]["label"],
                        "score" : reponse["features"][0]["properties"]["score"],
                        "postcode" : reponse["features"][0]["properties"]["postcode"],
                        "city" : reponse["features"][0]["properties"]["city"]
                    })
    else:
        liste.append({  "adresse" : adresse,
                        "label" : None,
                        "score" : None,
                        "postcode" : None,
                        "city" : None
                    })

print(liste) 
```

Etape 3 :

```python
df_api = pd.DataFrame(liste)
df_api
```

Etape 4 :

```python
print(f"Nombre d'adresses trouvées : {df_api[df_api["label"] != None]["label"].count()}")
print(f"Score le plus bas : {df_api[df_api["score"] == df_api["score"].min()]["adresse"].iloc[0]}") #l'adresse et le label renvoyé sont différents
#Le label renvoyé ne correspont pas toujours a ce que j'avais envoyé.
```
