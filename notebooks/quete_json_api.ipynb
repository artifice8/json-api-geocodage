Etape 1 :

```python
import json

with open("../data/villes.json", encoding="utf-8") as f:
    df = json.load(f)
```

```python
print(f"Il y a {len(df['villes'])} villes.")
print(f"La deuxieme ville est {df['villes'][1]['nom']}.")
print(f"A Syracuse, il y a {len(df['villes'][2]['restaurants'])} restaurants.")
print(f"Le premier restaurant de Palerme est {df['villes'][1]['restaurants'][1]['nom']}")
```

Etape 2 :

```python
for ville in df['villes']:
    print(f"A {ville['nom']}, il y a {len(ville['restaurants'])} restaurants.")
```

Etape 3 :

```python
import pandas as pd

data = pd.json_normalize(df, record_path = ['villes', 'restaurants'], meta = [['villes','nom'],['villes', 'region'],['villes', 'habitants']])
print(data)
```

Etape 4 :

```python
import statistics

data['note_moyenne'] = data['notes'].apply(lambda x: round(statistics.mean(x), 2))
print(data)
```

```python
print(f"La ville qui a la meilleure note moyenne est {data[data['note_moyenne'] == data['note_moyenne'].max()]['villes.nom'].iloc[0]}.")
```
