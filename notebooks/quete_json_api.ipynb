## Challenge 2 :

Etape 1 :

```python
import requests

def interroger(url, params, headers=None):
    try:
        r = requests.get(url, params=params, headers=headers, timeout=10)
        r.raise_for_status()
        return r.json()
    except requests.exceptions.Timeout:
        print("Trop lent, on abandonne cette requête")
    except requests.exceptions.HTTPError as e:
        print("Le serveur a répondu une erreur :", e)
    except requests.exceptions.RequestException as e:
        print("Problème réseau :", e)
    return None
```

```python
headers = {"User-Agent": "bootcamp-simplon-da/1.0 (jordanpiade@gmail.com)"}
url = "https://api-adresse.data.gouv.fr/search/"
r = requests.get(url, params={"q": "20 avenue de Ségur 75007 Paris"}, headers=headers)

print(r.status_code)
print(r.json()["features"][0]["properties"]["label"])
print(r.json()["features"][0]["properties"]["score"])
print(r.json())
```

Etape 2 :

```python
adresses = [
    "20 avenue de Ségur 75007 Paris",
    "Grand Place, Lille",
    "1 rue de la Paix, Paris",
    "place Bellecour, Lyon",
    "zzzzz qwerty 99999",
]

liste = []
for adresse in adresses:
    reponse = interroger(url, {"q": adresse, "limit": 1})
    if len(reponse["features"]) > 0:
        liste.append({  "adresse" : adresse,
                        "label" : reponse["features"][0]["properties"]["label"],
                        "score" : reponse["features"][0]["properties"]["score"],
                        "postcode" : reponse["features"][0]["properties"]["postcode"],
                        "city" : reponse["features"][0]["properties"]["city"]
                    })
    else:
        liste.append({  "adresse" : adresse,
                        "label" : None,
                        "score" : None,
                        "postcode" : None,
                        "city" : None
                    })

print(liste) 
```

Etape 3 :

```python
df_api = pd.DataFrame(liste)
df_api
```

## Challenge 3 :

Etape 1 :

```python
with open("../data/restaurants.json", encoding="utf-8") as f:
    df_resto = json.load(f)
```

```python
resto = pd.json_normalize(df_resto, record_path = ['restaurants'], meta = ['ville', 'pays'])
print(resto)
```

Etape 2 :

```python
resto['note_moyenne'] = resto['notes'].apply(lambda x: statistics.mean(x))
print(resto)
```

Etape 3 :

```python
import time

def geocode (adresse):
    url = "https://nominatim.openstreetmap.org/search"
    headers = {"User-Agent": "bootcamp-simplon-da/1.0 (jordanpiade@gmail.com)"}
    params = {"q": adresse, "format": "json", "limit": 1}
    try:
        r = requests.get(url, params=params, headers=headers, timeout=10)
        r.raise_for_status()
        resultats = r.json()
        if not resultats:
            return None, None
        return float(resultats[0]["lat"]), float(resultats[0]["lon"])
    except requests.exceptions.RequestException:
            return None, None
    finally:
        time.sleep(1)
```

```python
resto[['lat','lon']] = resto['adresse'].apply(geocode).apply(pd.Series)
print(resto)
```

Etape 4 :

```python
import folium

point = [resto.loc[0, 'lat'], resto.loc[0, 'lon']]          # latitude, longitude — Catane
m = folium.Map(location=point, zoom_start=15)
for nb in range(0, len(resto)):
    popup = f"{resto.loc[nb, 'nom']}, {resto.loc[nb, 'note_moyenne']}"
    folium.Marker(location=[resto.loc[nb,'lat'], resto.loc[nb,'lon']],
                  popup=popup).add_to(m)
m
```
